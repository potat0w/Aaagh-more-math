```python
import pandas as pd
import numpy as np

# Original Zoho data
df = pd.read_csv('/content/sample_data/Bill (demo).csv')

# QBO template
qbo = pd.read_csv('/content/qbo_import_template.csv')

print("Source shape:", df.shape)
print("QBO template shape:", qbo.shape)
```

```python
qbo_columns = qbo.columns.tolist()

print("QBO required columns:")
print(qbo_columns)
```

```python
output = pd.DataFrame()

output["*BillNo"] = df["Bill Number"]
output["*Supplier"] = df["Vendor Name"]

output["*BillDate"] = pd.to_datetime(
    df["Bill Date"]
).dt.strftime("%d/%m/%Y")

output["*DueDate"] = pd.to_datetime(
    df["Due Date"]
).dt.strftime("%d/%m/%Y")

output["Terms"] = df["Payment Terms Label"]

output["Location"] = ""
output["Memo"] = ""

output["*Account"] = df["Account"]

output["LineDescription"] = df["Item Name"].fillna("")

output["*LineAmount"] = df["Item Total"]

output["*LineTaxCode"] = df["Tax Name"].fillna("Tax Exempt")

output["LineTaxAmount"] = df["Tax Amount"].fillna(0)

output["Currency"] = df["Currency Code"]
```

```python
output.head()
```

```python
output = output[qbo_columns]

output.head()
```

```python
comparison = pd.DataFrame({
    "Source Bill Number": df["Bill Number"].head(10),
    "QBO BillNo": output["*BillNo"].head(10),

    "Source Vendor": df["Vendor Name"].head(10),
    "QBO Supplier": output["*Supplier"].head(10),

    "Source Item Total": df["Item Total"].head(10),
    "QBO Line Amount": output["*LineAmount"].head(10),

    "Source Tax": df["Tax Amount"].head(10),
    "QBO Tax": output["LineTaxAmount"].head(10)
})

comparison
```

```python
source_check = df.groupby("Bill Number").agg(
    SourceSubtotal=("SubTotal", "first"),
    SourceTotal=("Total", "first")
)
```

```python
output_check = output.groupby("*BillNo").agg(
    QBO_LineAmount=("*LineAmount", "sum"),
    QBO_Tax=("LineTaxAmount", "sum")
)
```

```python
check = source_check.join(output_check)
```

```python
check["QBO_Calculated_Total"] = (
    check["QBO_LineAmount"] + check["QBO_Tax"]
)
```

```python
check["Subtotal_Match"] = (
    check["SourceSubtotal"].round(2)
    ==
    check["QBO_LineAmount"].round(2)
)

check["Total_Match"] = (
    check["SourceTotal"].round(2)
    ==
    check["QBO_Calculated_Total"].round(2)
)

check.head(10)
```

```python
print("Total bills:", len(check))
print("Subtotal matched:", check["Subtotal_Match"].sum())
print("Total matched:", check["Total_Match"].sum())
```

```python
print("Missing tax names in source:",
      df["Tax Name"].isna().sum())

print("Tax Exempt used in output:",
      (output["*LineTaxCode"] == "Tax Exempt").sum())
```
