```python
DRIVE_TEXT2SQL_ZIP = "/content/drive/MyDrive/data/text2sql_for_drive.zip"
DRIVE_SPIDER_ZIP = "/content/drive/MyDrive/data/spider_data.zip"

from pathlib import Path
import zipfile, os, shutil
from google.colab import drive
drive.mount("/content/drive", force_remount=False)

TEXT2SQL_ZIP = Path(DRIVE_TEXT2SQL_ZIP)
SPIDER_ZIP = Path(DRIVE_SPIDER_ZIP)
assert TEXT2SQL_ZIP.is_file(), f"missing {TEXT2SQL_ZIP}"
assert SPIDER_ZIP.is_file(), f"missing {SPIDER_ZIP}"

WORK = Path("/content")
TEXT2SQL = WORK / "text2sql"

if TEXT2SQL.exists():
    shutil.rmtree(TEXT2SQL)
print("Unzipping", TEXT2SQL_ZIP, "->", WORK)
with zipfile.ZipFile(TEXT2SQL_ZIP) as z:
    z.extractall(WORK)

# zip is flat (src/, configs/, ...) not nested text2sql/
if not (TEXT2SQL / "src" / "common.py").is_file():
    TEXT2SQL.mkdir(parents=True, exist_ok=True)
    for name in ("src", "configs", "requirements.txt", "RUNBOOK.md", "colab_text2sql.ipynb", "assets", "logs"):
        src = WORK / name
        if src.exists() and not (TEXT2SQL / name).exists():
            shutil.move(str(src), str(TEXT2SQL / name))

assert (TEXT2SQL / "src" / "phase2_prep.py").is_file(), "re-upload text2sql_for_drive.zip to Drive"
os.chdir(TEXT2SQL)
print("src files:", sorted(p.name for p in (TEXT2SQL / "src").glob("*.py")))
print("cwd:", Path.cwd())
```

```python
from pathlib import Path

# write fast config
Path("configs/train_qlora_fast.yaml").write_text("""
model_name: Qwen/Qwen2.5-Coder-1.5B-Instruct
train_file: data/train.jsonl
output_dir: checkpoints/qlora_r16
seed: 42
max_train_rows: 2000

quantization:
  load_in_4bit: true
  bnb_4bit_quant_type: nf4
  bnb_4bit_use_double_quant: true
  compute_dtype: float16

lora:
  r: 16
  alpha: 32
  dropout: 0.05
  target_modules: [q_proj, k_proj, v_proj, o_proj, gate_proj, up_proj, down_proj]

training:
  epochs: 1
  lr: 2.0e-4
  batch_size: 8
  grad_accum: 2
  max_seq_len: 768
  warmup_ratio: 0.03
  lr_scheduler: cosine
  weight_decay: 0.0
  logging_steps: 10
  save_strategy: epoch
""".lstrip())

# support max_train_rows if missing
p = Path("src/phase3_train.py")
t = p.read_text()
if "max_train_rows" not in t:
    t = t.replace(
        'train = [encode(d) for d in load_jsonl(cfg["train_file"])]\nprint("train rows:", len(train))',
        'raw = load_jsonl(cfg["train_file"])\n'
        'if cfg.get("max_train_rows"):\n'
        '    raw = raw[: int(cfg["max_train_rows"])]\n'
        'train = [encode(d) for d in raw]\n'
        'print("train rows:", len(train))',
    )
    p.write_text(t)
    print("patched max_train_rows")

!python src/phase3_train.py configs/train_qlora_fast.yaml
```
