```python
# 8. Add periodic, resumable checkpoint saving to the existing training script
train_script = PROJECT / "src" / "phase3_train.py"
source = train_script.read_text()

if "save_steps=int(t.get(\"save_steps\", 25))" not in source:
    # Remove a pre-existing save_strategy argument, if present.
    source2 = re.sub(r"save_strategy\s*=\s*[^,\n]+,?\s*", "", source)

    # Insert checkpoint options immediately before gradient_checkpointing.
    marker = "gradient_checkpointing=True"
    if marker not in source2:
        raise RuntimeError(
            "Could not find 'gradient_checkpointing=True' in phase3_train.py. "
            "Open that file and adapt TrainingArguments manually."
        )

    checkpoint_args = (
        'save_strategy="steps", '
        'save_steps=int(t.get("save_steps", 25)), '
        'save_total_limit=int(t.get("save_total_limit", 3)), '
        'save_only_model=False, '
    )
    source2 = source2.replace(marker, checkpoint_args + marker, 1)
    train_script.write_text(source2)
    print("Patched phase3_train.py for periodic Drive checkpoints.")
else:
    print("Checkpoint options already appear to be patched.")

# Show the relevant part so you can verify it.
for i, line in enumerate(train_script.read_text().splitlines(), 1):
    if "TrainingArguments" in line or "save_strategy" in line or "save_steps" in line or "save_only_model" in line:
        print(f"{i}: {line}")
```
