```python
# 7a. Write a dedicated training config
CONFIG_PATH = PROJECT / "configs" / "train_drive_checkpoint.yaml"

config_text = f'''
model_name: Qwen/Qwen2.5-Coder-7B-Instruct
train_file: data/train.jsonl
output_dir: {DRIVE_CHECKPOINTS}
seed: 42
max_train_rows: 2000

quantization:
  load_in_4bit: true
  bnb_4bit_quant_type: nf4
  bnb_4bit_use_double_quant: true
  compute_dtype: float16

lora:
  r: 16
  alpha: 32
  dropout: 0.05
  target_modules: [q_proj, k_proj, v_proj, o_proj, gate_proj, up_proj, down_proj]

training:
  epochs: 1
  lr: 1.0e-4
  batch_size: 1
  grad_accum: 8
  max_seq_len: 512
  warmup_ratio: 0.03
  lr_scheduler: cosine
  weight_decay: 0.0
  logging_steps: 5
  save_strategy: steps
  save_steps: 25
  save_total_limit: 3
'''.lstrip()

CONFIG_PATH.write_text(config_text)
print(CONFIG_PATH.read_text())
```

```python
# 8. Add periodic, resumable checkpoint saving to the existing training script
train_script = PROJECT / "src" / "phase3_train.py"
source = train_script.read_text()

if "save_steps=int(t.get(\"save_steps\", 25))" not in source:
    # Remove a pre-existing save_strategy argument, if present.
    source2 = re.sub(r"save_strategy\s*=\s*[^,\n]+,?\s*", "", source)

    # Insert checkpoint options immediately before gradient_checkpointing.
    marker = "gradient_checkpointing=True"
    if marker not in source2:
        raise RuntimeError(
            "Could not find 'gradient_checkpointing=True' in phase3_train.py. "
            "Open that file and adapt TrainingArguments manually."
        )

    checkpoint_args = (
        'save_strategy="steps", '
        'save_steps=int(t.get("save_steps", 25)), '
        'save_total_limit=int(t.get("save_total_limit", 3)), '
        'save_only_model=False, '
    )
    source2 = source2.replace(marker, checkpoint_args + marker, 1)
    train_script.write_text(source2)
    print("Patched phase3_train.py for periodic Drive checkpoints.")
else:
    print("Checkpoint options already appear to be patched.")

# Show the relevant part so you can verify it.
for i, line in enumerate(train_script.read_text().splitlines(), 1):
    if "TrainingArguments" in line or "save_strategy" in line or "save_steps" in line or "save_only_model" in line:
        print(f"{i}: {line}")
```

```python
from pathlib import Path
import re

train_script = Path("/content/text2sql/src/phase3_train.py")
source = train_script.read_text()

# Remove every occurrence of checkpoint arguments from TrainingArguments
for arg in [
    "save_strategy",
    "save_steps",
    "save_total_limit",
    "save_only_model",
]:
    source = re.sub(
        rf'\b{arg}\s*=\s*(?:"[^"]*"|\'[^\']*\'|int\(t\.get\([^)]*\)\)|True|False|\d+)\s*,?\s*',
        "",
        source,
    )

# Insert checkpoint settings only once, before gradient_checkpointing
marker = "gradient_checkpointing=True"

if marker not in source:
    raise RuntimeError("gradient_checkpointing=True not found; no file written.")

checkpoint_args = (
    'save_strategy="steps", '
    'save_steps=int(t.get("save_steps", 25)), '
    'save_total_limit=int(t.get("save_total_limit", 3)), '
    'save_only_model=False, '
)

source = source.replace(marker, checkpoint_args + marker, 1)
train_script.write_text(source)

print("Fixed checkpoint arguments.")
print("\nTrainingArguments section:")
lines = source.splitlines()
for i, line in enumerate(lines):
    if "args = TrainingArguments(" in line:
        print("\n".join(f"{j+1}: {lines[j]}" for j in range(i, min(i+12, len(lines)))))
        break
```
