```python
import json, os, sqlite3, random

SPIDER = "/content/spider_data"   # jekhane unzip korsos
N_TRAIN = 3000                    # koto ta example diye shekhabo
N_TEST  = 300                     # koto ta question diye exam nibo
random.seed(42)                   # seed fix, jate prottekbar same row ashe

# Database er structure (CREATE TABLE ...) ber kore, eta model ke prompt e dei
def get_schema(db_path):
    con = sqlite3.connect(db_path)
    rows = con.execute("SELECT sql FROM sqlite_master "
                       "WHERE type='table' AND sql IS NOT NULL").fetchall()
    con.close()
    return "\n".join(r[0] for r in rows)

# Spider er json file porhe amader format e convert kore
def convert(src, n):
    data = json.load(open(f"{SPIDER}/{src}"))
    random.shuffle(data)
    out = []
    for d in data:
        db = d["db_id"]
        path = f"{SPIDER}/database/{db}/{db}.sqlite"   # accuracy mapar shomoy SQL ei file e chalabo
        if not os.path.exists(path): continue
        out.append({"id": len(out), "db_id": db, "db_path": path,
                    "schema": get_schema(path),
                    "question": d["question"],
                    "gold_sql": d["query"]})            # gold_sql = shothik answer
        if len(out) == n: break
    return out

train = convert("train_spider.json", N_TRAIN)   # shekhar jonno
test  = convert("dev.json", N_TEST)             # exam er jonno

# data/ folder e save
os.makedirs("/content/text2sql/data", exist_ok=True)
os.makedirs("/content/text2sql/results", exist_ok=True)
for name, rows in [("train", train), ("test", test)]:
    with open(f"/content/text2sql/data/{name}.jsonl", "w") as f:
        for r in rows:
            f.write(json.dumps(r, ensure_ascii=False) + "\n")

# Check: train ar test e same database ba question thakle cheating hoye jay
print(len(train), len(test))
print("db overlap (0 hote hobe):", len({r["db_id"] for r in train} & {r["db_id"] for r in test}))
print("question overlap:", len({r["question"] for r in train} & {r["question"] for r in test}))

# Phase 2 er info save (README te lagbe)
json.dump({"source": "Spider", "used_train": len(train), "used_test": len(test), "seed": 42,
           "split": "train_spider.json -> train, dev.json -> test (disjoint databases)",
           "examples": train[:3]},
          open("/content/text2sql/results/phase2_data_stats.json", "w"), indent=1)
```

```python
import json

tr = [json.loads(l) for l in open("/content/text2sql/data/train.jsonl")]
te = [json.loads(l) for l in open("/content/text2sql/data/test.jsonl")]

test_q = {r["question"] for r in te}
dup = [r for r in tr if r["question"] in test_q]
print("duplicate:", [r["question"] for r in dup])

tr = [r for r in tr if r["question"] not in test_q]
for i, r in enumerate(tr): r["id"] = i      # id abar sajai

with open("/content/text2sql/data/train.jsonl", "w") as f:
    for r in tr: f.write(json.dumps(r, ensure_ascii=False) + "\n")

print("train now:", len(tr), "| overlap:", len({r["question"] for r in tr} & test_q))

# Phase 2 stats update
st = json.load(open("/content/text2sql/results/phase2_data_stats.json"))
st["used_train"] = len(tr)
st["removed_duplicate_questions"] = len(dup)
json.dump(st, open("/content/text2sql/results/phase2_data_stats.json", "w"), indent=1)
```

```python
import json, yaml, sys
sys.path.insert(0, "/content/text2sql/src")
from transformers import AutoTokenizer
from common import prompt_text

%cd /content/text2sql
MODEL = "Qwen/Qwen2.5-1.5B-Instruct"     # Coder korte hole eta bodle de
MAXLEN = 1536

tok = AutoTokenizer.from_pretrained(MODEL)
rows = [json.loads(l) for l in open("data/train.jsonl")]

# original save kore rakhi
import shutil, os
if not os.path.exists("data/train_full.jsonl"):
    shutil.copy("data/train.jsonl", "data/train_full.jsonl")
rows = [json.loads(l) for l in open("data/train_full.jsonl")]

keep = []
for r in rows:
    n = len(tok(prompt_text(tok, r["schema"], r["question"]) + r["gold_sql"]).input_ids)
    if n <= MAXLEN: keep.append(r)
for i, r in enumerate(keep): r["id"] = i

with open("data/train.jsonl", "w") as f:
    for r in keep: f.write(json.dumps(r, ensure_ascii=False) + "\n")
print("raw:", len(rows), "-> kept:", len(keep), "| dropped:", len(rows) - len(keep))

# phase2 stats update
st = json.load(open("results/phase2_data_stats.json"))
st.update({"max_seq_len_filter": MAXLEN, "train_after_length_filter": len(keep),
           "dropped_too_long": len(rows) - len(keep)})
json.dump(st, open("results/phase2_data_stats.json", "w"), indent=1)

# training config update
cfg = yaml.safe_load(open("configs/train_qlora.yaml"))
cfg["model_name"] = MODEL
cfg["output_dir"] = "checkpoints/qlora_" + MODEL.split("/")[-1]
cfg["training"]["max_seq_len"] = MAXLEN
cfg["training"]["batch_size"] = 2        # lamba sequence, tai memory bachate
cfg["training"]["grad_accum"] = 8        # effective batch ekhono 16
yaml.safe_dump(cfg, open("configs/train_qlora.yaml", "w"), sort_keys=False)
print(open("configs/train_qlora.yaml").read())
```
